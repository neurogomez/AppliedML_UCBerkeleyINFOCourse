### 1.4 Simple Linear Regression

Estimate a linear regression of the median house value (the dependent variable) on the population (the independent variable), with no other control variables. Interpret the coefficients and standard errors. Based on this analysis, can you conclude anything about the causal effect of decreasing the population on the median housing value?

````python
import statsmodels.api as sm

# define the independent and dependent variables 
X = cal_df['Population']
y = cal_df['MedHouseVal']

# Add a constant term - ensures we have a constant term in the model
X = sm.add_constant(X)

# fit regression model
model = sm.OLS(y, X).fit()

# print the summary of OSL regression results 
print(model.summary())
````

*Write your response here.*

- The intercept is \$ 210,436 dollars as the median house value for a block population of zero, while the slope is -2.51 \$ /person. Since the slope is negative it indicates a trend to have higher median house values in smaller population blocks. The standard error is large for the constant parameter indicating there is high dispersion in the data set. The slope has a smaller standard error but it is still large relative to the absolute value of slope. 


- Although the there appears to be a correlation between the median house value and the size of the poulation block, we cannot conclude any causal effect of decreasing the population on the median housing value since the standard errors are non-negligble.
